```python
import pandas as pd
from sklearn.preprocessing import StandardScaler

# Load the dataset
df = pd.read_csv('churn_clean.csv')

# Drop columns that are identifiers or unrelated to prediction
df = df.drop(columns=[
    'Customer_id', 'CaseOrder', 'Interaction', 'UID', 'City', 'State', 'County', 'Zip',
    'Area', 'TimeZone', 'Job', 'Email',
    'Item1', 'Item2', 'Item3', 'Item4', 'Item5', 'Item6', 'Item7', 'Item8'
])

# Identify all remaining categorical columns
categorical_cols = df.select_dtypes(include=['object']).columns.tolist()

# Perform one-hot encoding
df = pd.get_dummies(df, columns=categorical_cols, drop_first=True)

# Set features (X) and target variable (y)
X = df.drop(columns=['MonthlyCharge'])
y = df['MonthlyCharge']

# Standardize the feature variables to have mean 0 and variance 1
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Save the cleaned dataset
df.to_csv('cleaned_churn_data.csv', index=False)
```

```python
from sklearn.model_selection import train_test_split
import pandas as pd

# Split the data (80% training, 20% testing)
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

# Convert scaled arrays back into DataFrames and add original column names
X_train_df = pd.DataFrame(X_train, columns=X.columns)
X_test_df = pd.DataFrame(X_test, columns=X.columns)

# Save X_train and X_test with column names
X_train_df.to_csv('X_train.csv', index=False)
X_test_df.to_csv('X_test.csv', index=False)

# Save y_train and y_test
y_train_df = pd.DataFrame(y_train)
y_test_df = pd.DataFrame(y_test)

y_train_df.to_csv('y_train.csv', index=False)
y_test_df.to_csv('y_test.csv', index=False)

#(Stojiljković)
```

```python
from sklearn.linear_model import Ridge
from sklearn.metrics import mean_squared_error, r2_score

# Initialize Ridge Regression model
model = Ridge(alpha=1.0)

# Train the model
model.fit(X_train, y_train)

# Predict on the test set
y_pred = model.predict(X_test)

# Evaluate model performance
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("Mean Squared Error (MSE):", mse)
print("R-squared Score:", r2)

#(Yadav)
```
